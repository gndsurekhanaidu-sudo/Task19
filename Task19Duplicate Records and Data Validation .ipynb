```python
import pandas as pd
```

```python
data = {
    'CustomerID': [1, 2, 3, 1, 4, 5, 6],
    'Name': ['Ravi', 'Priya', 'Kumar', 'Ravi',
             'Sita', 'Arun', 'John'],
    'Age': [20, 22, 150, 20, 21, -5, 25],
    'City': ['Vijayawada', 'Hyderabad', 'Chennai',
             'Vijayawada', 'Delhi', 'Mumbai', 'Chennai'],
    'Gender': ['Male', 'Female', 'Male', 'Male',
               'Female', 'Other', 'Male']
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.duplicated())
```

```python
print("Total duplicates:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print(df)
```

```python
# 1. Check missing values
print(df.isnull().sum())

# 2. Check duplicate records
print(df.duplicated().sum())

# 3. Check invalid ages
print(df[df['Age'] > 100])

# 4. Check negative ages
print(df[df['Age'] < 0])

# 5. Check valid gender categories
print(df[~df['Gender'].isin(['Male', 'Female'])])

# 6. Check missing customer IDs
print(df[df['CustomerID'].isnull()])
```

```python
print("Final Cleaned Dataset:")
print(df)
```

```python
df.to_csv("cleaned_customer.csv", index=False)
```
